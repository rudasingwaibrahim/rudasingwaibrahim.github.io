## Rappels

- $a$ et $b$ sont des entiers. $a$ est **un multiple de** $b$ s'il existe un entier $k$ tel que $a = k \times b$. On dit aussi que $b$ est **un diviseur** de $a$.
- En Python, pour des entiers naturels `a` et `b` (avec `b` non nul) :
  - `a // b` donne le **quotient** de la division euclidienne de `a` par `b` ;
  - `a % b` donne le **reste** de cette division.
- $a$ est un multiple de $b$ exactement quand le reste `a % b` vaut `0`.
- Un entier $n$ est **pair** s'il est un multiple de 2, **impair** sinon.

Exécute la cellule suivante et observe : $17 = 3 \times 5 + 2$.

```python
print("17 // 5 =", 17 // 5)
print("17 % 5 =", 17 % 5)
print("20 % 5 =", 20 % 5)
```
