- Le **déterminant** des vecteurs $\vec{u}\begin{pmatrix} x \\ y \end{pmatrix}$ et $\vec{v}\begin{pmatrix} x' \\ y' \end{pmatrix}$ est $\det(\vec{u}, \vec{v}) = x\,y' - y\,x'$.
- $\vec{u}$ et $\vec{v}$ sont **colinéaires** exactement quand $\det(\vec{u}, \vec{v}) = 0$.
- $A$, $B$, $C$ sont **alignés** exactement quand $\overrightarrow{AB}$ et $\overrightarrow{AC}$ sont colinéaires ; $(AB)$ et $(CD)$ sont **parallèles** exactement quand $\overrightarrow{AB}$ et $\overrightarrow{CD}$ sont colinéaires.
- Le vecteur $\overrightarrow{AB}$ a pour coordonnées $\begin{pmatrix} x_B - x_A \\ y_B - y_A \end{pmatrix}$.

- $A$, $B$, $C$ alignés $\iff \det(\overrightarrow{AB}, \overrightarrow{AC}) = 0$ ; $(AB) \parallel (CD) \iff \det(\overrightarrow{AB}, \overrightarrow{CD}) = 0$.
