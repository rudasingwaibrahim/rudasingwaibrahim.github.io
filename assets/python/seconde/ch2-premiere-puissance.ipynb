- `p = 1` puis `p = p * a` répété $n$ fois donne $a^n$ : c'est un accumulateur de produit.
- Pour trouver la **première** puissance qui dépasse une valeur, on utilise une boucle `while` : on multiplie **tant que** la condition d'arrêt n'est pas atteinte, et un compteur `n` compte les multiplications.
- Bien choisir la condition : « tant que $p \leq b$ » s'arrête au premier $p > b$ ; « tant que $p \geq b$ » s'arrête au premier $p < b$.
