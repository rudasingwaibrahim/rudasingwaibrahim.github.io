## Rappels

- $\sqrt{2}$ est le nombre positif dont le carré vaut 2. Il est **irrationnel** : son écriture décimale ne s'arrête jamais.
- Pour un nombre $a \geq 0$ : $a \leq \sqrt{2}$ exactement quand $a^2 \leq 2$. On peut donc comparer à $\sqrt{2}$ **sans** connaître $\sqrt{2}$.
- Un **encadrement d'amplitude** $10^{-n}$ de $\sqrt{2}$ s'écrit $a \leq \sqrt{2} < a + 10^{-n}$. En Python, $10^{-n}$ s'écrit `10 ** (-n)`.
- **Balayage** : on part de $a = 1$ (car $1^2 \leq 2$) et on avance de `pas` en `pas` **tant que** le nombre suivant $a + \text{pas}$ a un carré inférieur ou égal à 2.
