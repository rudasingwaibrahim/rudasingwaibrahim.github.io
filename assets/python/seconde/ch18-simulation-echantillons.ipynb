## 4. Moyenne et écart type d'une série de fréquences

On simule $N$ échantillons de taille $n$ : on obtient une série de $N$ fréquences $f_1, \ldots, f_N$.

**Lire.** La fonction `moyenne_frequences` ci-dessous renvoie la moyenne $m$ de cette série. Exécute la cellule.

```python
def moyenne_frequences(N, n, p):
    """Renvoie la moyenne des fréquences de N échantillons de taille n."""
    s = 0
    for k in range(N):
        s = s + frequence(n, p)
    return s / N
```

## 5. La loi des grands nombres

**Observer.** La cellule suivante simule des échantillons de taille $n$ de plus en plus grande (de 10 à 3000) pour l'expérience « obtenir un 6 », et place les points $(n\,;\,f)$. La droite horizontale est $y = \dfrac{1}{6}$.

```python
import matplotlib.pyplot as plt

for n in range(10, 3001, 50):
    plt.plot(n, frequence(n, 1 / 6), "b.")
plt.axhline(1 / 6, color="red")
plt.xlabel("taille n de l'échantillon")
plt.ylabel("fréquence des 6")
plt.show()
```

**Loi des grands nombres** (version du programme) : lorsque $n$ est grand, sauf exception, la fréquence observée est proche de la probabilité.

- D'un échantillon à l'autre, la fréquence **fluctue** ; plus $n$ est grand, plus les fréquences sont concentrées autour de $p$.
