## Le principe

On choisit au hasard un point $M(x\,;\,y)$ dans le carré $[0\,;\,1] \times [0\,;\,1]$ : $x$ et $y$ sont deux nombres choisis au hasard, **indépendamment**, entre 0 et 1. La probabilité que $M$ tombe dans un domaine du carré est égale à l'**aire** de ce domaine (l'aire du carré vaut 1).

## Bilan

- Pour un point $M(x\,;\,y)$ tiré au hasard dans le carré unité, $x$ et $y$ sont tirés **séparément** (indépendamment).
- La probabilité que $M$ soit dans un domaine est l'aire du domaine ; la **fréquence** observée sur $n$ points en est une estimation.
- Aire sous la parabole sur $[0\,;\,1]$ : environ $\frac{1}{3}$ ; quart de disque : $\frac{\pi}{4}$, donc $\pi \approx 4 \times$ fréquence.
