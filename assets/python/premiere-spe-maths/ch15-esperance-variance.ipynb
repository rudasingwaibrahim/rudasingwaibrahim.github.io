## Rappels

Une variable aléatoire $X$ prend les valeurs $x_1, \ldots, x_n$ avec les probabilités $p_1, \ldots, p_n$. On représente sa loi par **deux listes** de même longueur : `X = [x_1, ..., x_n]` et `P = [p_1, ..., p_n]`. Attention : en Python, les indices commencent à 0 : `X[0]` est $x_1$.

- **Espérance** : $E(X) = p_1 x_1 + p_2 x_2 + \cdots + p_n x_n$.
- **Variance** : $V(X) = p_1 (x_1 - E(X))^2 + \cdots + p_n (x_n - E(X))^2$.
- **Écart type** : $\sigma(X) = \sqrt{V(X)}$.
- `len(X)` est le nombre d'éléments de `X` ; `for i in range(len(X)):` parcourt les indices $0, 1, \ldots, n - 1$.
