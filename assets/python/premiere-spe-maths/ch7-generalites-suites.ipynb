## Bilan

- Formule explicite : un seul calcul. Récurrence : une boucle qui applique $n$ fois la relation à partir de $u_0$.
- De $u_0$ à $u_n$, il y a $n + 1$ termes.
- Un **accumulateur** (`S = S + terme`) calcule une somme de termes.
- Un **algorithme de seuil** utilise une boucle `while` : il permet de conjecturer une limite, mais ne la démontre pas.
