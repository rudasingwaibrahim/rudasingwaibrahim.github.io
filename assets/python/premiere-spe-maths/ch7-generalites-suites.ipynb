- Le carré de `n` s'écrit `n ** 2` (ou `n * n`).

## 4. Algorithme de seuil

Il semble que les termes $v_n$ se rapprochent de 6 en restant inférieurs à 6. On cherche le **plus petit** entier $n$ tel que l'écart $6 - v_n$ soit strictement inférieur à un nombre $e > 0$ donné.

**Interpréter.** Plus $e$ est petit, plus le rang trouvé est grand, mais on trouve toujours un rang à partir duquel $v_n$ est aussi proche de 6 qu'on le souhaite. On **conjecture** que la suite $(v_n)$ a pour limite 6.

**Une limite infinie.** Pour la suite $w_n = n^2$, les termes dépassent n'importe quel nombre $A$, aussi grand soit-il.

**À toi.** Complète `seuil_carre(A)`, qui renvoie le plus petit entier naturel $n$ tel que $n^2 > A$. On continue **tant que** $n^2 \leq A$.

```python
def seuil_carre(A):
    """Renvoie le plus petit entier naturel n tel que n² > A."""
    n = 0
    while ... <= A:
        n = n + 1
    return n
```

```python
# Vérification
def test_seuil_carre():
    r = seuil_carre(1000)
    assert r == 32, f"31² = 961 ≤ 1000 et 32² = 1024 > 1000 : seuil_carre(1000) doit renvoyer 32. Ta fonction renvoie {montre(r)}."
    r = seuil_carre(10 ** 6)
    assert r != 1000, "1000² = 1 000 000 n'est pas strictement supérieur à 10⁶. On continue tant que n² ≤ A : la condition est n ** 2 <= A."
    assert r == 1001, f"seuil_carre(10 ** 6) doit renvoyer 1001. Ta fonction renvoie {montre(r)}."
    assert seuil_carre(0) == 1, "0² = 0 n'est pas strictement supérieur à 0, mais 1² = 1 l'est : seuil_carre(0) doit renvoyer 1."

verifier(test_seuil_carre)
```

On conjecture que la suite $(n^2)$ a pour limite $+\infty$. Enfin, une suite peut n'avoir **aucune** limite : les termes de $(-1)^n$ valent alternativement $1$ et $-1$, ils ne se rapprochent d'aucun nombre.

## 6. Les suites de Syracuse

On choisit un entier $c \geq 1$. On part de $u_0 = c$ puis, pour passer d'un terme au suivant : si le terme est pair, on le divise par 2 ; s'il est impair, on le multiplie par 3 et on ajoute 1. On s'arrête quand on obtient 1.

Exemple avec $c = 6$ : 6, 3, 10, 5, 16, 8, 4, 2, 1.

**À toi.** Complète les deux `...` de `syracuse(c)`, qui renvoie la liste des termes, de $c$ jusqu'au premier 1. Utilise `//` pour que les termes restent des entiers.

Pour $c = 27$, le chemin est très long et monte jusqu'à 9232 avant de redescendre ! Pour tous les entiers de départ testés à ce jour, on finit par atteindre 1, mais **personne n'a encore démontré** que c'est vrai pour tout entier $c \geq 1$ : c'est la **conjecture de Syracuse**.

## Bilan

- Formule explicite : un seul calcul. Récurrence : une boucle qui applique $n$ fois la relation à partir de $u_0$.
- De $u_0$ à $u_n$, il y a $n + 1$ termes.
- Un **accumulateur** (`S = S + terme`) calcule une somme de termes.
- Un **algorithme de seuil** utilise une boucle `while` : il permet de conjecturer une limite, mais ne la démontre pas.
