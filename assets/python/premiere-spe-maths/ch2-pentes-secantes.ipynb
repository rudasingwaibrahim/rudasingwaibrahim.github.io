## Bilan

- Le taux de variation `(f(a + h) - f(a)) / h` est la pente de la sécante.
- Une liste se construit **par ajouts successifs** (`L.append(...)` dans une boucle) ou **en compréhension** (`[... for k in range(...)]`).
- Si les pentes se rapprochent d'un nombre quand $h$ tend vers 0, on **conjecture** que c'est $f'(a)$ ; on le **démontre** par le calcul.
