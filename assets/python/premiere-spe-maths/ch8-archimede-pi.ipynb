## Le principe

Un polygone régulier à $N$ côtés **inscrit** dans un cercle de rayon 1 a un périmètre $p_N = 2N\sin\left(\dfrac{\pi}{N}\right)$ (cours : chaque côté mesure $2\sin\left(\dfrac{\pi}{N}\right)$). Le polygone **exinscrit** (dont les côtés touchent le cercle) a un périmètre $q_N = \dfrac{p_N}{\cos(\pi/N)}$.

Le cercle a pour périmètre $2\pi$ et il est « coincé » entre les deux polygones : $p_N < 2\pi < q_N$, donc
$$\frac{p_N}{2} < \pi < \frac{q_N}{2}.$$

Archimède (IIIe siècle avant notre ère) part de l'hexagone ($N = 6$) et double quatre fois le nombre de côtés, jusqu'à 96.

## Bilan

- $\dfrac{p_N}{2} = N\sin\left(\dfrac{\pi}{N}\right) < \pi < \dfrac{q_N}{2} = \dfrac{N\sin(\pi/N)}{\cos(\pi/N)}$.
- En doublant le nombre de côtés, on resserre l'encadrement : c'est la méthode d'Archimède.
- Une boucle `while` donne le premier $N$ pour lequel l'amplitude est inférieure à une précision donnée.
