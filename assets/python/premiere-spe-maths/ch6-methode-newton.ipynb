## Le principe

$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}.$$

## Bilan

- Méthode de Newton : $x \leftarrow x - \dfrac{f(x)}{f'(x)}$, à condition que $f'(x) \neq 0$ et que l'on parte assez près de la solution (cas favorable).
- Dans les cas favorables, la convergence est très rapide : quelques étapes suffisent pour une grande précision.
- Avec $f(x) = x^2 - 2$, on retrouve la méthode de Héron : $x \leftarrow \frac{1}{2}\left(x + \frac{2}{x}\right)$.
