## Le principe de la méthode d'Euler

La fonction exponentielle vérifie $\exp' = \exp$ et $\exp(0) = 1$. Pour $h$ proche de 0, l'approximation affine donne
$$\exp(x + h) \approx \exp(x) + h\,\exp'(x) = \exp(x) + h\,\exp(x) = (1 + h)\exp(x).$$

On part du point $(0\,;\,1)$ et on avance de proche en proche avec le pas $h$ : si $y$ est l'ordonnée du point d'abscisse $x$, le point suivant a pour abscisse $x + h$ et pour ordonnée $y + h\,y$.
