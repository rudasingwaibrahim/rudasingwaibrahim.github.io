## Rappels

- **Inégalité de Bienaymé-Tchebychev** : pour une variable $X$ d'espérance $\mu$ et de variance $V$, et tout $\delta > 0$, $P(|X - \mu| \geq \delta) \leq \dfrac{V}{\delta^2}$.
- **Inégalité de concentration** : si $M_n$ est la moyenne d'un échantillon de taille $n$ d'une variable d'espérance $\mu$ et de variance $V$, alors $P(|M_n - \mu| \geq \delta) \leq \dfrac{V}{n\,\delta^2}$.
- Pour $S_n \sim \mathcal{B}(n\,;\,p)$ : $E(S_n) = np$ et $V(S_n) = np(1-p)$. Avec $\delta = \sqrt{n}$ : $P(|S_n - np| \geq \sqrt{n}) \leq p(1-p)$.
