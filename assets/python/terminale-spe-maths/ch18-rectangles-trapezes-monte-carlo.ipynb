## Bilan

- Rectangles : $h \sum f(x_k)$ ; milieux : $h \sum f\left(x_k + \frac{h}{2}\right)$ ; trapèzes : $h \sum \frac{f(x_k) + f(x_{k+1})}{2}$.
- L'erreur des rectangles est de l'ordre de $\frac{1}{n}$, celle des milieux et des trapèzes de l'ordre de $\frac{1}{n^2}$.
- Monte-Carlo : aire du rectangle $\times$ fréquence des points sous la courbe.
