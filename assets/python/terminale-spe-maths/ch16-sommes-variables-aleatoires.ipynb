## Bilan

- Une variable binomiale $\mathcal{B}(n\,;\,p)$ est la somme de $n$ variables de Bernoulli indépendantes : $E = np$, $V = np(1-p)$.
- L'espérance d'une somme est la somme des espérances ; pour des variables **indépendantes**, la variance d'une somme est la somme des variances.
- La moyenne $M_n$ d'un échantillon a la même espérance que $X$, mais une variance $n$ fois plus petite : $V(M_n) = \dfrac{V(X)}{n}$.
