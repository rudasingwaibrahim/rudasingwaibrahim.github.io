## Le principe

Pour une fonction $y$ dérivable et un pas $h$ petit : $y(x + h) \approx y(x) + h\,y'(x)$. On part du point $(x_0\,;\,y_0)$ donné par la condition initiale et on avance de proche en proche :
$$x_{k+1} = x_k + h, \qquad y_{k+1} = y_k + h \times y'(x_k),$$
où $y'(x_k)$ est donné par l'équation différentielle. Les points $(x_k\,;\,y_k)$ forment une ligne brisée proche de la courbe de la solution.

## Bilan

- Méthode d'Euler : $y_{k+1} = y_k + h \times y'(x_k)$, où $y'(x_k)$ est donné par l'équation : $f(x_k)$ pour $y' = f$, $a y_k + b$ pour $y' = ay + b$.
- La ligne brisée obtenue se rapproche de la courbe de la solution quand $h$ diminue.
- Le cours donne la solution exacte de $y' = ay + b$ : $x \mapsto C e^{ax} - \frac{b}{a}$.
