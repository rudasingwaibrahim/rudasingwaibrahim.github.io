## Le principe

Pour $A > 0$ et $B > 0$ : $\ln\sqrt{AB} = \dfrac{1}{2}\ln(AB) = \dfrac{\ln A + \ln B}{2}$. La **moyenne géométrique** $\sqrt{AB}$ a donc pour logarithme la **moyenne arithmétique** des logarithmes.

Soit $x$ avec $1 \leq x < e$. On part de $A = 1$ ($\ln A = 0$) et $B = e$ ($\ln B = 1$), de sorte que $A \leq x \leq B$. À chaque étape, on calcule $C = \sqrt{AB}$ et $\ln C = \dfrac{\ln A + \ln B}{2}$, puis on remplace l'une des bornes par $C$ pour que $x$ reste entre $A$ et $B$. Les bornes se rapprochent de $x$, et $\ln A$, $\ln B$ se rapprochent de $\ln x$.

En Python : `from math import sqrt, e`. Les variables `lA`, `lB`, `lC` contiennent $\ln A$, $\ln B$, $\ln C$.

- L'algorithme de Briggs resserre $[A\,;\,B]$ autour de $x$ ; $\ln A$ et $\ln B$ se rapprochent de $\ln x$.
- Les propriétés $\ln(e^k y) = k + \ln y$ et $\ln\frac{1}{x} = -\ln x$ ramènent tout $x > 0$ au cas $1 \leq x < e$.
