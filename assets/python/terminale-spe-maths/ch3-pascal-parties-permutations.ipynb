## Rappels

- $n! = 1 \times 2 \times \cdots \times n$ pour $n \geq 1$, et $0! = 1$.
- $\dbinom{n}{k}$ est le nombre de parties à $k$ éléments d'un ensemble à $n$ éléments. **Relation de Pascal** : pour $1 \leq k \leq n - 1$, $\dbinom{n}{k} = \dbinom{n-1}{k-1} + \dbinom{n-1}{k}$, avec $\dbinom{n}{0} = \dbinom{n}{n} = 1$.
- Le nombre de $k$-uplets d'éléments distincts d'un ensemble à $n$ éléments est $n(n-1)\cdots(n-k+1)$ ; le nombre de permutations est $n!$.
- On représente un ensemble fini par une **liste** d'éléments distincts, par exemple `E = ["a", "b", "c", "d", "e"]`.

## 5. Tirer une permutation au hasard

## Bilan

- Le triangle de Pascal se construit ligne par ligne : chaque coefficient intérieur est la somme des deux coefficients situés au-dessus.
- Pour les parties à 2 éléments, on impose $i < j$ : chaque partie est comptée une seule fois ($\binom{n}{2}$ parties).
