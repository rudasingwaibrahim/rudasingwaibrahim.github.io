## Bilan

- Les premiers termes d'une suite récurrente permettent de **conjecturer** une formule ; on la **démontre** par récurrence.
- Un algorithme de seuil s'écrit avec `while` : on calcule le terme suivant et on augmente $n$ **tant que** la condition d'arrêt n'est pas atteinte.
- Une recherche de seuil illustre la définition d'une limite ; elle ne la démontre pas.
