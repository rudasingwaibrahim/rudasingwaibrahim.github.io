## Rappels : le symbole $\Sigma$

$\displaystyle\sum_{k=1}^{n} k^2$ se lit « somme pour $k$ allant de 1 à $n$ de $k^2$ » : c'est $1^2 + 2^2 + \cdots + n^2$.

| Dans l'écriture $\displaystyle\sum_{k=1}^{n} k^2$ | Dans l'algorithme | En Python |
|---|---|---|
| la somme vaut 0 avant d'ajouter le premier terme | **initialisation** de l'accumulateur | `S = 0` |
| l'indice $k$ va de 1 à $n$ | **compteur** de la boucle | `for k in range(1, n + 1):` |
| on ajoute le terme $k^2$ | **accumulateur** | `S = S + k ** 2` |
| le dernier terme est celui de rang $k = n$ | **sortie de boucle** | après le tour $k = n$, on renvoie `S` |

Attention : `range(1, n + 1)` donne les entiers de 1 à $n$ ; `range(n)` donne ceux de 0 à $n - 1$.

**Exemple corrigé.** Exécute cette cellule : elle calcule $\displaystyle\sum_{k=1}^{n} k = 1 + 2 + \cdots + n$ et compare avec la formule $\dfrac{n(n+1)}{2}$.

```python
def somme_entiers(n):
    """Renvoie 1 + 2 + ... + n."""
    S = 0                      # initialisation de l'accumulateur
    for k in range(1, n + 1):  # compteur : k va de 1 à n
        S = S + k              # accumulateur : on ajoute le terme k
    return S                   # sortie de boucle : on renvoie la somme

for n in [1, 5, 10, 100]:
    print(n, somme_entiers(n), n * (n + 1) // 2)
```
