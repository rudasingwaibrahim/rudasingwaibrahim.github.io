- Dans un schéma de Bernoulli de $n$ épreuves, $\dbinom{n}{k}$ est le nombre de chemins de l'arbre qui comportent $k$ succès. Ces nombres se lisent dans le **triangle de Pascal** : chaque nombre intérieur est la somme des deux nombres situés au-dessus (formule de Pascal).
- Si $X$ suit la loi binomiale $\mathcal{B}(n\,;\,p)$ : $P(X = k) = \dbinom{n}{k} p^k (1 - p)^{n - k}$ et $E(X) = np$.

- En Python, `T[i]` est l'élément d'indice `i` de la liste `T` (le premier a l'indice 0).

## 4. Simulation et diagramme en bâtons

```python
from random import random

def nb_succes(n, p):
    """Renvoie le nombre de succès dans un échantillon de taille n d'une loi de Bernoulli de paramètre p."""
    s = 0
    for i in range(n):
        if random() < p:
            s = s + 1
    return s

def frequences(N, n, p):
    """Simule N échantillons ; renvoie la liste des fréquences de 0, 1, ..., n succès."""
    eff = [0 for k in range(n + 1)]
    for j in range(N):
        s = nb_succes(n, p)
        eff[s] = ...
    return [... for e in eff]
```
